```python
import pandas as pd
```

```python
df = pd.read_csv('used_cars_dataset.csv')
print(f"Original shape: {df.shape}")
df.head()
```

 Clean AskPrice
Strip the ₹ symbol and commas, then convert to a numeric (float) column.

```python
df['Price'] = (
    df['AskPrice']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(float)
)
df.drop(columns=['AskPrice'], inplace=True)
df['Price'].head()
```

```python
df
```

**Clean kmDriven**
Strip the  km text and commas, then convert to float.

```python
df['kmDriven'] = (
    df['kmDriven']
    .str.replace(' km', '', regex=False)
    .str.replace(',', '', regex=False)
    .str.strip()
    .astype(float)
)
df['kmDriven'].head()
```

```python
# Detect missing values
df.isnull().sum()
```

Handle missing values
kmDriven has 88 missing values — fill them with the median of the column

```python
km_median = df['kmDriven'].median()
df['kmDriven'] = df['kmDriven'].fillna(km_median)
print(f"Missing values in kmDriven: {df['kmDriven'].isna().sum()}")
```

**Standardise categorical columns**

FuelType has inconsistent values like 'hybrid' and 'Hybrid/CNG' — unify them
Owner has lowercase values like 'first', 'second' — capitalise them
Strip any extra whitespace from Brand and Model

```python
df['FuelType'] = df['FuelType'].str.strip().str.title()
df['FuelType'] = df['FuelType'].replace({'Hybrid/Cng': 'CNG/Hybrid', 'Hybrid': 'CNG/Hybrid'})
df['Owner'] = df['Owner'].str.strip().str.capitalize()
df['Brand'] = df['Brand'].str.strip()
df['model'] = df['model'].str.strip()
df.rename(columns={'model': 'Model'}, inplace=True)
print(df['FuelType'].unique())
print(df['Owner'].unique())
```

```python
print(df['Year'].min())
print(df['Year'].max())
```

**Year — 1990 to 2024**

```python
df = df[(df['Year'] >= 1990) & (df['Year'] <= 2024)]
```

```python
#checking min and max km
print(df['kmDriven'].min())
print(df['kmDriven'].max())
```

```python
df[df['kmDriven'] == 0]
```

```python
df[df['kmDriven'] > 500000]['kmDriven'].value_counts()
```

```python
#discard 0km and above 5lac km
df = df[(df['kmDriven'] > 0) & (df['kmDriven'] <= 500000)]
```

```python
print(df['Price'].min())
print(df['Price'].max())
```

```python
df[df['Price'] >2000000]['Price'].value_counts()
```

```python
#discard less than 20k and higher than 2cr
df = df[(df['Price'] >= 50000) & (df['Price'] <= 20000000)]
```

Drop duplicates

```python
df.drop_duplicates(inplace=True)
```

Reset Index

```python
df.reset_index(drop=True, inplace=True)
```

**Verify**

```python
print(f"Cleaned shape: {df.shape}")
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
```

**saving**

```python
df.to_csv('used_cars_cleaned.csv', index=False)
print("Cleaned dataset saved!")
```
